# Extract, Transformation, and Load System

### Importing libraries

```python
from pathlib import Path
import pandas as pd
from sqlalchemy import create_engine
```

### Opening the CSV files and uploading them to the 'raw' schema in pgAdmin 4.

```python
# 1. Define the file path using Path
DATA_DIR = Path("../data/raw")

# Note: The line below is unnecessary since you are reading the files inside the loop.
# dfs = pd.read_csv(DATA_DIR)

# 2. Create the PostgreSQL database connection
# Format: postgresql://USERNAME:PASSWORD@HOST:PORT/DATABASE_NAME
engine = create_engine(
    "postgresql://postgres:malbaecleia@localhost:5432/retailsales"
)

SCHEMA_NAME = "raw"

# 3. Loop through each .csv file in the folder
for archive in DATA_DIR.glob("*.csv"):
    df = pd.read_csv(archive)

    # Uses the file name itself (without the .csv extension) as the table name
    table_name = archive.stem

    df.to_sql(
        name=table_name,
        con=engine,
        schema=SCHEMA_NAME,
        if_exists="append",
        index=False,
    )

    print(f"File '{archive.name}' saved in table '{table_name}'.")
```
